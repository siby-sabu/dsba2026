## Vegetation Districts

```python
import pandas as pd
from pathlib import Path

df_vegetation = pd.read_pickle("data/raw/pickle/df_vegetation_2016_2026.pkl")

# Optional: standardize column names
df_vegetation.columns = df_vegetation.columns.str.lower()

# View data
print(df_vegetation.head())
print(df_vegetation.shape)
```

```python
# Keep only the district column and drop duplicate values
df_vegetation_districts = df_vegetation[["district_name"]].drop_duplicates()
df_vegetation_districts["district_name"]= df_vegetation_districts["district_name"].str.lower().str.replace(" ", "_")
df_vegetation_districts
```

```python
import pandas as pd

# 1. Standardize column names for easy comparison
df_master = df_lat_long_districts[["district", "district_id"]].rename(columns={"district": "district_master"})
df_crop = df_cropyield_districts[["district_name"]].rename(columns={"district_name": "district_crop"})
df_precip = df_precipitation_districts[["district_name"]].rename(columns={"district_name": "district_precip"})
df_veg = df_vegetation_districts[["district_name"]].rename(columns={"district_name": "district_veg"})

# 2. Add presence flags
df_crop["in_crop"] = True
df_precip["in_precip"] = True
df_veg["in_veg"] = True

# 3. Check direct matches against the Master list
unmatched_crop = set(df_crop["district_crop"]) - set(df_master["district_master"])
unmatched_precip = set(df_precip["district_precip"]) - set(df_master["district_master"])
unmatched_veg = set(df_veg["district_veg"]) - set(df_master["district_master"])

print("--- Unmatched Counts ---")
print(f"Crop districts not in Master: {len(unmatched_crop)}")
print(f"Precipitation districts not in Master: {len(unmatched_precip)}")
print(f"Vegetation districts not in Master: {len(unmatched_veg)}")
```

```python
# Inspect 15 random unmatched districts from Crop and Precip
print("--- Sample Unmatched Crop Districts ---")
print(list(unmatched_crop)[:15])

print("\n--- Sample Unmatched Precip Districts ---")
print(list(unmatched_precip)[:15])

print("\n--- Sample Master Districts ---")
print(list(df_master["district_master"])[:15])
```

```python
shared_unmatched = unmatched_crop.intersection(unmatched_precip)
print(f"Districts unmatched in BOTH Crop and Precip: {len(shared_unmatched)}")
```

```python
from rapidfuzz import process, fuzz
import pandas as pd

master_list = df_master["district_master"].dropna().unique().tolist()
all_unmatched = sorted(list(unmatched_crop.union(unmatched_precip).union(unmatched_veg)))

records = []
for name in all_unmatched:
    match, score, _ = process.extractOne(name, master_list, scorer=fuzz.token_sort_ratio)
    records.append({
        "raw_district": name,
        "matched_master": match,
        "score": round(score, 1),
        "category": "High Confidence (>=85)" if score >= 85 else ("Review (70-84)" if score >= 70 else "Low Match (<70)")
    })

df_match_summary = pd.DataFrame(records).sort_values(by="score", ascending=False)

# Check distribution
print(df_match_summary["category"].value_counts())
```

```python
# 1. Export mapping summary to CSV
df_match_summary.to_csv("district_mapping_check.csv", index=False)
print("Saved to district_mapping_check.csv for visual verification!")
```
