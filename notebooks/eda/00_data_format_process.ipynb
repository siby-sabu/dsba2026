```python
import pandas as pd
import numpy as np
```

```python
from os import mkdir
from pathlib import Path
project_root = next(p for p in [Path.cwd(), *Path.cwd().parents] if p.name == "dsba2026")
#pickle_folder_path = "../../data/raw/pickle"
pickle_folder_path = project_root / "data" / "raw" / "pickle"
Path(pickle_folder_path).mkdir(parents=True, exist_ok=True)
```

```python
#  Percipitation
current = Path.cwd()
project_root = current if (current / "CLAUDE.md").exists() else current.parent.parent
pickle_folder_path = project_root / "data" / "raw" / "pickle"
```

```python
# Crop Yield

crop_yield_folder_path = project_root / "data" / "raw" / "crop_yield"
pickle_folder_path = (
    project_root / "data" / "raw" / "pickle" / "df_crop_yield_2016_2026.pkl"
)

df_list = []
for file_path in crop_yield_folder_path.iterdir():
    if file_path.is_file and file_path.suffix == ".csv":
        df = pd.read_csv(file_path)
        df_list.append(df)

crop_yield_df = pd.concat(df_list)
crop_yield_df.to_pickle(pickle_folder_path)
```

```python
# Geography

geography_folder_path = project_root / "data" / "raw" / "geography"
pickle_folder_path = project_root / "data" / "raw" / "pickle" / "df_geography.pkl"

df_list = []
for file_path in geography_folder_path.iterdir():
    if file_path.is_file and file_path.suffix == ".csv":
        df = pd.read_csv(file_path)
        df_list.append(df)

geography_df = pd.concat(df_list)
geography_df.to_pickle(pickle_folder_path)
```
