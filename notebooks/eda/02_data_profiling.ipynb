```python
#  Update the datatypes for the datasets
#  Crop Yield dataset
# Pick Year 1:
# 	Kharif: Monsoon crop sown in June/July and harvested in Sept/Oct of Year 1
# 	Autumn: A sub-season of Kharif harvested completely in the late months of Year 1.
# 	Total / Whole Year: Per national reporting conventions, long-duration crops or full-year totals are mapped to the start year of the agricultural cycle.
# Pick Year 2 :
# 	Rabi: Winter crop sown in Oct/Nov of Year 1 but grown, matured, and harvested in early Year 2.
# 	Winter: Heavily overlaps with Rabi, extending into the early months of Year 2.
# 	Summer: Sown late in the cycle (often called Zaid) and harvested completely in the hot months of Year 2.


from matplotlib import axis

crop_yield_df["year1"] = crop_yield_df["year"].str.split("-").str[0]
crop_yield_df["year2"] = crop_yield_df["year"].str.split("-").str[1]
conditions = [
    crop_yield_df["season"].isin(["Kharif", "Autumn", "Total", "Whole Year"]),
    crop_yield_df["season"].isin(["Rabi", "Winter", "Summer"]),
]
choices = [
    pd.to_datetime(crop_yield_df["year1"]),
    pd.to_datetime(crop_yield_df["year2"]),
]
crop_yield_df["year"] = pd.to_datetime(
    np.select(conditions, choices, default=crop_yield_df["year1"])
)
crop_yield_df.drop(["year1", "year2"], axis=1, inplace=True)
crop_col_convert = {
    col: "category" for col in ["state", "district", "crop_name", "crop_type", "season"]
}
crop_yield_df = crop_yield_df.astype(crop_col_convert)

geo_col_covert = {col: "category" for col in ["state", "district"]}
geography_df = geography_df.astype(geo_col_covert)

precipit_col_convert = {col: "category" for col in ["state", "district"]}
precipitation_df = precipitation_df.astype(precipit_col_convert)
precipitation_df["date"] = pd.to_datetime(precipitation_df["Date"])


vegetation_col_convert = {
    col: "category" for col in ["district", "state", "quality_status"]
}
vegetation_df = vegetation_df.astype(vegetation_col_convert)
vegetation_date_cols = ["date", "period_end"]
vegetation_df[vegetation_date_cols] = vegetation_df[vegetation_date_cols].apply(
    pd.to_datetime
)
```
