```python
from pathlib import Path

import pandas as pd


current = Path.cwd()
project_root = current if (current / "CLAUDE.md").exists() else current.parent.parent

pickle_folder_path = project_root / "data" / "processed"

crop_yield_df = pd.read_pickle(pickle_folder_path/"crop_yield.pkl")
geography_df = pd.read_pickle(pickle_folder_path/"geography.pkl")
precipitation_df = pd.read_pickle(pickle_folder_path/"precipitation.pkl")
vegetation_df = pd.read_pickle(pickle_folder_path/"vegetation.pkl")
```

```python
def profile_schema(df, dataset_name):
    print(f"\n{'='*60}")
    print(f"{dataset_name} Dataset")
    print(f"\n{'='*60}")
    print(f"Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
    print("\n Data Types:")
    print(df.dtypes)
    print("\n Non-null values")
    df.info()
```

```python
for df,name in [(crop_yield_df, 'Crop Yield'),(geography_df, 'Geography'), (precipitation_df, "Precipitation"), (vegetation_df, "Vegetation") ]:
    profile_schema(df, name)
```

```python
# Data type analysis
for df_name, df, columns in [('crop_yield_df',crop_yield_df, ['year']), ('precipitation_df', precipitation_df,['Date']),('vegetation_df',vegetation_df,['date','period_end','year','month','crop_mask_year','previous_composite_date'])]:
    print(f"\n{'='*60}")
    print(df_name)
    print(f"\n{'='*60}")
    print(df[columns].head())
```
