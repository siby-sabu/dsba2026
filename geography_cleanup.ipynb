## Extract

```python
import pandas as pd
import os
 
df_lat_long = pd.read_pickle("data/raw/pickle/df_geography.pkl")

df_lat_long.columns = df_lat_long.columns.str.lower()
df_lat_long
```

```python
df_lat_long["district"]= df_lat_long["district"].str.lower().str.replace(" ", "_")
df_lat_long["state"]= df_lat_long["state"].str.lower().str.replace(" ", "_")
df_lat_long
```

## Transform

```python
# # Step 1: Condense lat-long data by district using mean
# df_lat_long_condensed = (
#  df_lat_long
# .groupby("district", as_index=False)[["latitude", "longitude"]]
# .mean()
# )

# Step 1: Condense lat-long data by district and preserve district ID using mean
df_lat_long_condensed = (
    df_lat_long
    .groupby(["district", "state", "ind_adm2_id"], as_index=False)[["latitude", "longitude"]]
    .mean()
) 
df_lat_long_condensed = df_lat_long_condensed.rename(columns={"ind_adm2_id": "district_id"})
df_lat_long_condensed
```
