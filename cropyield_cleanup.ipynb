## Extract

```python
import pandas as pd
from pathlib import Path

df_cropyield = pd.read_pickle("data/raw/pickle/df_crop_yield_2016_2026.pkl")

# Optional: standardize column names
df_cropyield.columns = df_cropyield.columns.str.lower()

# View data
print(df_cropyield.head())
print(df_cropyield.shape)
```

## Transform

```python
df_cropyield["state_name"]= df_cropyield["state_name"].str.lower().str.replace(" ", "_")
df_cropyield["district_name"]= df_cropyield["district_name"].str.lower().str.replace(" ", "_")
df_cropyield["crop_name"]= df_cropyield["crop_name"].str.lower().str.replace(" ", "_")
df_cropyield["crop_type"]= df_cropyield["crop_type"].str.lower().str.replace(" ", "_")
df_cropyield["season"]= df_cropyield["season"].str.lower().str.replace(" ", "_")
df_cropyield["area_unit"]= df_cropyield["area_unit"].str.lower().str.replace(" ", "_")
df_cropyield["production_unit"]= df_cropyield["production_unit"].str.lower().str.replace(" ", "_")
df_cropyield["yield_unit"]= df_cropyield["yield_unit"].str.lower().str.replace(" ", "_")
df_cropyield
```

```python
import pandas as pd

# Clean and create base year
df_cropyield["year_base"] = (
 df_cropyield["year"]
.astype(str)
.str.strip()
.str.split("-")
.str[0]
.astype(int)
)
df_cropyield
```

```python
df_cropyield_filtered = df_cropyield[['crop_code', 'crop_name', 'crop_type']].drop_duplicates()
df_cropyield_filtered
```

```python
df_cropyield_filtered1 = df_cropyield[['year_base', 'district_name', 'crop_code', 'season', 'area', 'production', 'yield']]
df_cropyield_filtered1
```

```python
# Count the number of null values in each column
null_counts = df_cropyield_filtered1.isnull().sum()

print("--- Null Values Per Column ---")
print(null_counts)
```

```python
# List of columns where you want to ensure NO null values exist
columns_to_check = ["year_base", "district_name", "crop_code", "season","production"]

# Filter the DataFrame to keep only rows that are not null in those columns
df_cropyield_filtered1 = df_cropyield_filtered1.dropna(subset=columns_to_check)
df_cropyield_filtered1
```

```python
# Count the total number of duplicate rows across the selected columns
duplicate_count = df_cropyield_filtered1.duplicated(
    subset=["year_base", "district_name", "crop_code", "season"]
).sum()

print(f"Total duplicate rows: {duplicate_count}")
```

```python
# 1. Identify all occurrences of duplicates and store them in a new DataFrame
df_duplicates = df_cropyield_filtered1[
    df_cropyield_filtered1.duplicated(
        subset=["year_base", "district_name", "crop_code", "season"], 
        keep=False
    )
].sort_values(by=["district_name", "year_base", "crop_code", "season"])

# 2. Print the total count and display the DataFrame
print(f"Stored {len(df_duplicates)} duplicate rows in 'df_duplicates'.")
print("\n--- First 10 duplicate rows ---")
df_duplicates.head(10)
```
