# Step 6 : Data Integration

```python
df_combined = df_precipitation_filtered.copy()
```

```python
import pandas as pd

# Keep only one NDVI value per year-month
ndvi_lookup = (
 df_vegetation[["year", "month", "ndvi_mean"]]
.dropna(subset=["year", "month"])
.drop_duplicates(subset=["year", "month"])
.set_index(["year", "month"])["ndvi_mean"]
)

# Map NDVI into df_combined without a large merge
df_combined["ndvi"] = pd.MultiIndex.from_frame(
 df_combined[["year", "month"]]
).map(ndvi_lookup)

df_combined.head()
```

```python
# Step 1: Condense lat-long data by district using mean
df_lat_long_condensed = (
 df_lat_long
.groupby("district", as_index=False)[["latitude", "longitude"]]
.mean()
)

df_lat_long_condensed
```

```python
# Step 2: Create lookup Series for mapping
latitude_lookup = df_lat_long_condensed.set_index("district")["latitude"]
longitude_lookup = df_lat_long_condensed.set_index("district")["longitude"]

latitude_lookup
longitude_lookup
```

```python
import pandas as pd

# Step 3: Remove old latitude/longitude columns if already present
df_combined = df_combined.drop(columns=["latitude", "longitude"], errors="ignore")

# Step 4: Map latitude and longitude into df_combined
df_combined["latitude"] = df_combined["district_name"].map(latitude_lookup)
df_combined["longitude"] = df_combined["district_name"].map(longitude_lookup)

# View result
df_combined.head()
```

## Feature Engineering

```python
import numpy as np
import pandas as pd

def classify_vegetation(ndvi):
    if pd.isna(ndvi):
        return np.nan
    elif ndvi <= 0.1:
        return "Non-vegetated (Water/Snow/Cloud)"
    elif ndvi <= 0.15:
        return "Bare soil or rock"
    elif ndvi <= 0.5:
        return "Moderate/Low Vegetation"
    elif ndvi <= 1.0:
        return "Dense Vegetation"
    else:
        return np.nan

df_combined["Vegetation"] = (
df_combined["ndvi"].apply(classify_vegetation)
)

df_combined
```

```python
 # Ensure data is sorted chronologically by date
df_combined = df_combined.sort_values(by="date")

# Create the 1-month lag column group by the district
df_combined["precipitation_lag_1m"] = df_combined.groupby("district_name")["precipitation_mm"].shift(1)

df_combined
```

```python
df_combined["precipitation_change"] = df_combined["precipitation_mm"].diff()
df_combined["ndvi_change"] = df_combined["ndvi"].diff()
df_combined["precipitation_roll_2m"] = df_combined["precipitation_mm"].rolling(2).mean()


df_combined
```

```python
df_combined["is_pre_monsoon"] = df_combined["month"].isin([3, 4, 5]).astype(int)

df_combined
```
