```python
import pandas as pd

# Keep only one NDVI value per year-month
ndvi_lookup = (
 df_vegetation[["year", "month", "ndvi_mean"]]
.dropna(subset=["year", "month"])
.drop_duplicates(subset=["year", "month"])
.set_index(["year", "month"])["ndvi_mean"]
)

# Map NDVI into df_combined without a large merge
df_combined["ndvi"] = pd.MultiIndex.from_frame(
 df_combined[["year", "month"]]
).map(ndvi_lookup)

df_combined.head()
```

## Feature Engineering

```python
import numpy as np
import pandas as pd

def classify_vegetation(ndvi):
    if pd.isna(ndvi):
        return np.nan
    elif ndvi <= 0.1:
        return "Non-vegetated (Water/Snow/Cloud)"
    elif ndvi <= 0.15:
        return "Bare soil or rock"
    elif ndvi <= 0.5:
        return "Moderate/Low Vegetation"
    elif ndvi <= 1.0:
        return "Dense Vegetation"
    else:
        return np.nan

df_ppt_ap_eg_filt_mon["Vegetation"] = (
df_ppt_ap_eg_filt_mon["ndvi"].apply(classify_vegetation)
)

df_ppt_ap_eg_filt_mon
```

```python
 # Ensure data is sorted chronologically by date
df_ppt_ap_eg_filt_mon = df_ppt_ap_eg_filt_mon.sort_values(by="date")

# Create the 1-month lag column group by the district
df_ppt_ap_eg_filt_mon["precipitation_lag_1m"] = df_ppt_ap_eg_filt_mon.groupby("district_name")["precipitation_mm"].shift(1)

df_ppt_ap_eg_filt_mon
```

```python
df_ppt_ap_eg_filt_mon["precipitation_change"] = df_ppt_ap_eg_filt_mon["precipitation_mm"].diff()
df_ppt_ap_eg_filt_mon["ndvi_change"] = df_ppt_ap_eg_filt_mon["ndvi"].diff()
df_ppt_ap_eg_filt_mon["precipitation_roll_2m"] = df_ppt_ap_eg_filt_mon["precipitation_mm"].rolling(2).mean()


df_ppt_ap_eg_filt_mon
```

```python
df_ppt_ap_eg_filt_mon["is_pre_monsoon"] = df_ppt_ap_eg_filt_mon["month"].isin([3, 4, 5]).astype(int)

df_ppt_ap_eg_filt_mon
```
