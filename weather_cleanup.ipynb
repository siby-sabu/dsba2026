## Extract

```python
import pandas as pd
from pathlib import Path

df_precipitation = pd.read_pickle("data/raw/pickle/df_precipitation_2016_2026.pkl")

# Optional: standardize column names
df_precipitation.columns = df_precipitation.columns.str.lower()

# View data
print(df_precipitation.head())
print(df_precipitation.shape)
```

## Transform

```python
df_precipitation["state_name"]= df_precipitation["state_name"].str.lower().str.replace(" ", "_")
df_precipitation["district_name"]= df_precipitation["district_name"].str.lower().str.replace(" ", "_")
df_precipitation
```

```python
df_precipitation=df_precipitation[df_precipitation['district_name'].notnull()]
df_precipitation=df_precipitation[df_precipitation['state_name'].notnull()]
df_precipitation=df_precipitation[df_precipitation['date'].notnull()]
df_precipitation=df_precipitation[df_precipitation['precipitation_mm']>=0]
```

```python
import pandas as pd

# Clean and create year and month
# Convert date column to datetime
df_precipitation["date"] = pd.to_datetime(df_precipitation["date"])

# Extract year and month
df_precipitation["year"] = df_precipitation["date"].dt.year
df_precipitation["month"] = df_precipitation["date"].dt.month

df_precipitation
```

```python
# Aggregate directly to get exactly 1 row per unique combination of district, year, and month
df_precipitation_filtered = (
    df_precipitation
    .groupby(["district_name", "year", "month"], as_index=False)["precipitation_mm"]
    .mean()
    .rename(columns={"precipitation_mm": "precipitation_mm_mean"})
)

df_precipitation_filtered
```

## Load

```python
df_precipitation_filtered.to_pickle(r"data\raw\processed_ram\fact_precipitation.pkl")
```
